```python
#choosing the optimal chunk size
raw_path = Path("../data/raw/goodreads_interactions_fantasy_paranormal.json")


reader = pd.read_json(
    raw_path,
    lines=True,
    chunksize=250_000
)

test_chunk = next(reader)

full_memory = (
    test_chunk.memory_usage(deep=True).sum()
    / 1024**2
)

print(f"Full chunk: {full_memory:.2f} MB")
```

```python
keep_columns = [
    "user_id",
    "book_id",
    "review_id",
    "rating",
    "is_read",
    "date_added"
]

reduced_chunk = test_chunk[keep_columns].copy()

reduced_memory = (
    reduced_chunk.memory_usage(deep=True).sum()
    / 1024**2
)

print(f"Full chunk: {full_memory:.2f} MB")
print(f"Reduced cols: {reduced_memory:.2f} MB")
```

```python
#test
CHUNK_SIZE = 250_000

test_reader = pd.read_json(
    raw_path,
    lines=True,
    chunksize=CHUNK_SIZE
)

test_chunk = next(test_reader)

test_reduced = test_chunk[keep_columns].copy()

test_reduced["rating"] = pd.to_numeric(
    test_reduced["rating"],
    downcast="integer"
)

output_dir = Path("../data/interim/interactions_parts")
output_dir.mkdir(parents=True, exist_ok=True)

test_path = (
    output_dir
    / "interactions_test.parquet"
)

test_reduced.to_parquet(
    test_path,
    index=False,
    compression="zstd"
)

print("Test file:", test_path)
print("File exists:", test_path.exists())
print(
    "File size:",
    round(test_path.stat().st_size / 1024**2, 2),
    "MB"
)
```

```python
test_loaded = pd.read_parquet(test_path)

print("Before saving:", test_reduced.shape)
print("After loading:", test_loaded.shape)

print("\nColumns identical:")
print(test_reduced.columns.tolist() == test_loaded.columns.tolist())
```

```python
dtype_comparison = pd.DataFrame({
    "before": test_reduced.dtypes.astype(str),
    "after": test_loaded.dtypes.astype(str)
})

dtype_comparison
```

```python
null_comparison = pd.DataFrame({
    "before": test_reduced.isna().sum(),
    "after": test_loaded.isna().sum()
})

null_comparison
```

```python
pd.testing.assert_frame_equal(
    test_reduced.reset_index(drop=True),
    test_loaded.reset_index(drop=True),
    check_dtype=True
)

print("Test passed: values and data types are identical.")
```

```python
import math

SAMPLE_LINES = 100_000
sample_bytes = 0
sample_count = 0

with raw_path.open("rb") as file:
    for _ in range(SAMPLE_LINES):
        line = file.readline()

        if not line:
            break

        sample_bytes += len(line)
        sample_count += 1

average_line_size = sample_bytes / sample_count
file_size = raw_path.stat().st_size

estimated_rows = round(file_size / average_line_size)
estimated_chunks = math.ceil(estimated_rows / CHUNK_SIZE)

seconds_per_chunk = 12
estimated_minutes = estimated_chunks * seconds_per_chunk / 60

print(f"Average line size: {average_line_size:.0f} bytes")
print(f"Estimated rows: {estimated_rows:,}")
print(f"Estimated chunks: {estimated_chunks:,}")
print(f"Estimated conversion time: {estimated_minutes * 1.2:.1f} min")
```

```python
import pyarrow.parquet as pq

part_files = sorted(output_dir.glob("part_*.parquet"))

rows_per_file = [
    pq.ParquetFile(file).metadata.num_rows
    for file in part_files
]

first_schema = pq.read_schema(part_files[0])

schemas_identical = all(
    pq.read_schema(file).equals(first_schema)
    for file in part_files
)

total_size_gb = sum(
    file.stat().st_size
    for file in part_files
) / 1024**3

print("Files:", len(part_files))
print("Total rows:", f"{sum(rows_per_file):,}")
print("First chunk:", f"{rows_per_file[0]:,}")
print("Last chunk:", f"{rows_per_file[-1]:,}")
print("Schemas identical:", schemas_identical)
print("Total Parquet size:", f"{total_size_gb:.2f} GB")
```
